```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Charger les données
df = pd.read_csv("/content/base.csv")

# Conserver uniquement les variables numériques
num_cols = df.select_dtypes(include=[np.number])

# Matrice de corrélation de Spearman
cor_matrix = num_cols.corr(method="spearman")

# Taille de la figure
plt.figure(figsize=(12, 10))

# Heatmap
sns.heatmap(
    cor_matrix,
    annot=True,          # afficher les coefficients
    fmt=".4f",           # 4 décimales
    cmap="coolwarm",     # bleu-blanc-rouge
    vmin=-1,
    vmax=1,
    square=True,
    linewidths=0.5,
    cbar_kws={"label": "Corrélation"}
)

plt.title("Corrélation entre variables numériques", fontsize=14)
plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)

plt.tight_layout()
plt.show()
```

### Test de Mann-Whitney U pour toutes les paires de variétés

Nous allons maintenant étendre le test de Mann-Whitney U pour comparer la variable 'ssm' entre toutes les paires uniques de variétés.


```python
from scipy.stats import mannwhitneyu
import itertools

# Obtenir toutes les variétés uniques
unique_varieties = df['variete'].unique()

# Préparer une liste pour stocker les résultats
results = []

# Générer toutes les paires uniques de variétés
for v1, v2 in itertools.combinations(unique_varieties, 2):
    # Séparer les groupes pour la variété 1 et la variété 2
    g1 = df.loc[df['variete'] == v1, 'ssm'].dropna()
    g2 = df.loc[df['variete'] == v2, 'ssm'].dropna()

    # S'assurer qu'il y a suffisamment de données pour le test
    if len(g1) > 0 and len(g2) > 0:
        # Exécuter le test de Mann-Whitney U
        stat, p = mannwhitneyu(g1, g2, alternative='two-sided')
        results.append({'Variété 1': v1, 'Variété 2': v2, 'Statistique U': stat, 'p-value': p})
    else:
        results.append({'Variété 1': v1, 'Variété 2': v2, 'Statistique U': None, 'p-value': None})
        print(f"Attention : Pas assez de données pour comparer {v1} et {v2}.")

# Convertir les résultats en DataFrame pour une meilleure visualisation
results_df = pd.DataFrame(results)

print("Résultats des tests de Mann-Whitney U pour 'ssm' entre les paires de variétés :")
display(results_df)
```

#### Interprétation des résultats et considération sur les comparaisons multiples

Le tableau ci-dessus présente les statistiques U et les p-valeurs pour chaque paire de variétés comparées. Une p-value faible (typiquement inférieure à 0.05) suggère une différence statistiquement significative dans les distributions de 'ssm' entre les deux variétés comparées.

Il est important de noter que lorsque plusieurs tests statistiques sont effectués simultanément, la probabilité d'obtenir un résultat "significatif" par pur hasard augmente (le problème des comparaisons multiples). Pour une analyse rigoureuse, des méthodes de correction des p-valeurs (comme la correction de Bonferroni ou de Benjamini-Hochberg) peuvent être appliquées pour ajuster ces seuils de signification. Cependant, pour cette étape d'exploration, nous présentons les p-valeurs brutes.

### Test de Kruskal-Wallis pour la variable 'ssm' entre toutes les variétés

Le test de Kruskal-Wallis est une alternative non-paramétrique à l'ANOVA à un facteur, utilisée pour déterminer s'il existe des différences significatives entre les médianes de trois groupes ou plus sur une variable continue ou ordinale. Nous allons l'appliquer à la variable 'ssm' pour toutes les variétés.

```python
from scipy.stats import kruskal

# Obtenir toutes les variétés uniques (nous savons qu'il n'y en a que deux d'après le test précédent, mais le code est générique)
unique_varieties_kruskal = df['variete'].unique()

# Préparer une liste de groupes de données 'ssm' par variété
groups = [df.loc[df['variete'] == variety, 'ssm'].dropna() for variety in unique_varieties_kruskal]

# S'assurer qu'il y a au moins deux groupes pour le test de Kruskal-Wallis
if len(groups) > 1:
    # Exécuter le test de Kruskal-Wallis
    # L'opérateur '*' déballe la liste 'groups' en arguments individuels pour la fonction kruskal
    h_statistic, p_value_kruskal = kruskal(*groups)

    print(f"Test de Kruskal-Wallis pour 'ssm' entre les variétés :")
    print(f"Statistique H = {h_statistic:.4f}")
    print(f"p-value = {p_value_kruskal:.4e}")
else:
    print("Moins de deux variétés uniques trouvées. Le test de Kruskal-Wallis nécessite au moins deux groupes.")
    print(f"Variétés trouvées : {unique_varieties_kruskal}")
```

#### Interprétation du test de Kruskal-Wallis

Le résultat du test de Kruskal-Wallis fournit une statistique H et une p-value. Une p-value faible (typiquement inférieure à 0.05) indique qu'il existe une différence statistiquement significative dans les distributions de la variable 'ssm' entre au moins deux des groupes de variétés comparés. Si la p-value est élevée, cela suggère qu'il n'y a pas de différence significative.

Dans notre cas, comme précédemment identifié, la colonne 'variete' ne contient que deux variétés uniques. Le test de Kruskal-Wallis peut être appliqué à deux groupes, et ses résultats seront cohérents avec le test de Mann-Whitney U pour deux groupes, mais il est généralement plus informatif lorsqu'il y a trois groupes ou plus.

### Test de Mann-Whitney U pour toutes les variables numériques entre 'Amina1' et 'Fanaye1'

Nous allons maintenant effectuer le test de Mann-Whitney U pour comparer la distribution de chaque variable numérique entre les variétés 'Amina1' et 'Fanaye1'.

```python
from scipy.stats import mannwhitneyu

# Définir les deux variétés à comparer
variety1 = 'Amina1'
variety2 = 'Fanaye1'

# Sélectionner toutes les colonnes numériques du DataFrame
# Exclure la colonne 'variable_cluster' si elle a été ajoutée et n'est pas une mesure
numerical_columns = df.select_dtypes(include=[np.number]).columns.tolist()
if 'variable_cluster' in numerical_columns:
    numerical_columns.remove('variable_cluster')

# Préparer une liste pour stocker les résultats de tous les tests
all_variable_test_results = []

for col in numerical_columns:
    # Extraire les données pour la colonne actuelle pour chaque variété
    data_v1 = df.loc[df['variete'] == variety1, col].dropna()
    data_v2 = df.loc[df['variete'] == variety2, col].dropna()

    # S'assurer qu'il y a suffisamment de données pour le test
    if len(data_v1) > 0 and len(data_v2) > 0:
        try:
            # Exécuter le test de Mann-Whitney U
            stat, p_value = mannwhitneyu(data_v1, data_v2, alternative='two-sided')
            all_variable_test_results.append({
                'Variable': col,
                'Variété 1': variety1,
                'Variété 2': variety2,
                'Statistique U': stat,
                'p-value': p_value
            })
        except ValueError as e:
            all_variable_test_results.append({
                'Variable': col,
                'Variété 1': variety1,
                'Variété 2': variety2,
                'Statistique U': None,
                'p-value': None,
                'Erreur': str(e)
            })
            print(f"Attention : Erreur lors du test pour la variable {col}: {e}")
    else:
        all_variable_test_results.append({
            'Variable': col,
            'Variété 1': variety1,
            'Variété 2': variety2,
            'Statistique U': None,
            'p-value': None,
            'Erreur': 'Pas assez de données pour le test'
        })
        print(f"Attention : Pas assez de données dans la variable {col} pour comparer {variety1} et {variety2}.")

# Convertir les résultats en DataFrame pour une meilleure visualisation
all_results_df = pd.DataFrame(all_variable_test_results)

print(f"Résultats des tests de Mann-Whitney U pour toutes les variables numériques entre {variety1} et {variety2} :")
display(all_results_df)
```

```python
import pandas as pd
import numpy as np
from scipy.stats import mannwhitneyu

variables = [
    "temp","humidity","ec","ph","nitrogen","phosphorus",
    "potassium","salinity","ndvi","ndre","savi","cire",
    "ndwi","lai","fapar","biomasse_cum","temp_min_c",
    "temp_moy_c","temp_max_c","hum_rel_min_pct",
    "hum_rel_moy_pct","hum_rel_max_pct","vent_ms",
    "rad_sol_jcm2","rad_sol_mjm2","vv","vh","rvi",
    "ssm","das"
]

resultats = []

for var in variables:

    g1 = df.loc[df["variete"]=="Amina1", var].dropna()
    g2 = df.loc[df["variete"]=="Fanaye1", var].dropna()

    n1 = len(g1)
    n2 = len(g2)

    U, p = mannwhitneyu(g1, g2, alternative="two-sided")

    # Rank-biserial correlation
    rbc = (2 * U) / (n1 * n2) - 1

    # Interprétation
    abs_rbc = abs(rbc)

    if abs_rbc < 0.10:
        effet = "Négligeable"
    elif abs_rbc < 0.30:
        effet = "Faible"
    elif abs_rbc < 0.50:
        effet = "Modéré"
    else:
        effet = "Fort"

    resultats.append([
        var, U, p, rbc, effet
    ])

resultats = pd.DataFrame(
    resultats,
    columns=[
        "Variable",
        "U",
        "p_value",
        "Rank_biserial_r",
        "Effet"
    ]
)

resultats = resultats.sort_values("p_value")

print(resultats)
```

### Test de Kruskal-Wallis pour toutes les variables numériques entre les variétés

```python
from scipy.stats import kruskal

# Obtenir toutes les variétés uniques
unique_varieties = df['variete'].unique()

# Sélectionner toutes les colonnes numériques du DataFrame
numerical_columns = df.select_dtypes(include=[np.number]).columns.tolist()
if 'variable_cluster' in numerical_columns:
    numerical_columns.remove('variable_cluster')

# Préparer une liste pour stocker les résultats de tous les tests
kruskal_results = []

for col in numerical_columns:
    # Préparer une liste de groupes de données par variété pour la colonne actuelle
    groups_for_kruskal = [df.loc[df['variete'] == variety, col].dropna() for variety in unique_varieties]

    # Filtrer les groupes vides (ce qui peut arriver si une variété n'a pas de données pour une colonne)
    groups_for_kruskal = [g for g in groups_for_kruskal if len(g) > 0]

    if len(groups_for_kruskal) > 1:
        try:
            # Exécuter le test de Kruskal-Wallis
            h_statistic, p_value_kruskal = kruskal(*groups_for_kruskal)
            kruskal_results.append({
                'Variable': col,
                'Statistique H': h_statistic,
                'p-value': p_value_kruskal
            })
        except ValueError as e:
            kruskal_results.append({
                'Variable': col,
                'Statistique H': None,
                'p-value': None,
                'Erreur': str(e)
            })
            print(f"Attention : Erreur lors du test de Kruskal-Wallis pour la variable {col}: {e}")
    else:
        kruskal_results.append({
            'Variable': col,
            'Statistique H': None,
            'p-value': None,
            'Erreur': 'Moins de deux groupes avec des données pour le test'
        })
        print(f"Attention : Moins de deux groupes avec des données pour le test de Kruskal-Wallis pour la variable {col}.")

# Convertir les résultats en DataFrame pour une meilleure visualisation
kruskal_results_df = pd.DataFrame(kruskal_results)

# Trier par p-value pour voir les différences les plus significatives en premier
kruskal_results_df = kruskal_results_df.sort_values(by='p-value')

print("Résultats du test de Kruskal-Wallis pour toutes les variables numériques :")
display(kruskal_results_df)
```

#### Interprétation des résultats du test de Kruskal-Wallis

Le tableau ci-dessus présente les résultats du test de Kruskal-Wallis pour chaque variable numérique, comparant les distributions entre les variétés 'Amina1' et 'Fanaye1'.

*   **Statistique H :** La statistique de test de Kruskal-Wallis.
*   **p-value :** La probabilité d'observer une telle différence (ou une différence plus extrême) si l'hypothèse nulle (pas de différence entre les distributions) était vraie. Une p-value faible (généralement < 0.05) indique une différence statistiquement significative entre les distributions des variables pour au moins deux des groupes (ici, les deux variétés).

Comme pour le test de Mann-Whitney U, il est important de considérer que le fait de réaliser de nombreux tests peut augmenter la probabilité d'obtenir des résultats faussement positifs. Pour une analyse rigoureuse, une correction des comparaisons multiples (par exemple, Bonferroni ou Benjamini-Hochberg) pourrait être appliquée aux p-values.

### Test de Kruskal-Wallis pour toutes les variables numériques groupées par 'site' et 'variete'

### Test post-hoc de Dunn avec correction de Benjamini-Hochberg par 'site_variete'

```python
import scikit_posthocs as sp

# Sélectionner toutes les colonnes numériques du DataFrame
numerical_columns = df.select_dtypes(include=[np.number]).columns.tolist()
# Exclure la colonne 'variable_cluster' et 'das' (qui peut être un indice de temps ou un ordinal)
if 'variable_cluster' in numerical_columns:
    numerical_columns.remove('variable_cluster')
if 'das' in numerical_columns:
    numerical_columns.remove('das')

# Préparer une liste pour stocker les résultats des tests post-hoc
dunn_results_list = []

for col in numerical_columns:
    # Vérifier s'il y a plus d'un groupe avec des données pour cette variable
    # (le test de Dunn nécessite au moins 2 groupes)
    unique_groups = df.loc[df[col].notna(), 'site_variete'].unique()

    if len(unique_groups) > 1:
        try:
            # Effectuer le test post-hoc de Dunn pour la variable actuelle
            # La correction 'fdr_bh' est pour Benjamini-Hochberg
            # 'bonferroni' est une autre option si l'on veut une correction plus stricte
            dunn_result = sp.posthoc_dunn(df, val_col=col, group_col='site_variete', p_adjust='fdr_bh')

            # Renommer les colonnes et index pour plus de clarté
            dunn_result.columns = [f"p_value_{c}" for c in dunn_result.columns]
            dunn_result.index = [f"p_value_{r}" for r in dunn_result.index]

            dunn_results_list.append({
                'Variable': col,
                'Dunn_Test_Results': dunn_result
            })
        except Exception as e:
            dunn_results_list.append({
                'Variable': col,
                'Dunn_Test_Results': None,
                'Erreur': str(e)
            })
            print(f"Attention : Erreur lors du test de Dunn pour la variable {col}: {e}")
    else:
        dunn_results_list.append({
            'Variable': col,
            'Dunn_Test_Results': None,
            'Erreur': 'Moins de deux groupes avec des données pour le test de Dunn'
        })
        print(f"Attention : Moins de deux groupes avec des données pour le test de Dunn pour la variable {col}.")

print("Résultats des tests post-hoc de Dunn (p-values corrigées Benjamini-Hochberg) :")
# Afficher les résultats pour chaque variable
for res in dunn_results_list:
    print(f"\n--- Variable : {res['Variable']} ---")
    if res['Dunn_Test_Results'] is not None:
        display(res['Dunn_Test_Results'])
    else:
        print(f"  {res['Erreur']}")
```

## Analyse Multivariée : PCA (Analyse en Composantes Principales)

L'ACP est une technique de réduction de dimensionnalité qui transforme un ensemble de variables corrélées en un ensemble de variables non corrélées appelées composantes principales. Cela permet de visualiser les données de grande dimension dans un espace de dimension inférieure (généralement 2 ou 3 dimensions) tout en maximisant la variance expliquée. C'est également une étape préliminaire utile avant d'autres analyses multivariées comme PERMANOVA.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import seaborn as sns

# Sélectionner les colonnes numériques pertinentes pour la PCA
# Exclure 'variable_cluster' et 'das' car ce ne sont pas des mesures directes pour la PCA
pca_numerical_columns = df.select_dtypes(include=[np.number]).columns.tolist()
if 'variable_cluster' in pca_numerical_columns:
    pca_numerical_columns.remove('variable_cluster')
if 'das' in pca_numerical_columns:
    pca_numerical_columns.remove('das')

X = df[pca_numerical_columns].copy()

# Gérer les valeurs manquantes : pour la PCA, nous allons simplement supprimer les lignes avec des NaN
# Une imputation pourrait être envisagée pour des jeux de données plus grands ou avec beaucoup de manquants
X_clean = X.dropna()

# Mettre à jour le DataFrame principal pour inclure seulement les lignes avec des données complètes pour la PCA
df_pca = df.loc[X_clean.index].copy()

# Standardisation des données (très important pour la PCA)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_clean)

# Appliquer la PCA
pca = PCA()
X_pca = pca.fit_transform(X_scaled)

# Créer un DataFrame pour les résultats de la PCA
pca_df = pd.DataFrame(data=X_pca, columns=[f'PC{i+1}' for i in range(X_pca.shape[1])])

# Ajouter les informations de 'site_variete' au DataFrame PCA pour la visualisation
pca_df['site_variete'] = df_pca['site_variete'].values

# Afficher la variance expliquée par chaque composante
explained_variance_ratio = pca.explained_variance_ratio_
print("Variance expliquée par composante principale :\n", explained_variance_ratio)
print(f"Variance cumulée expliquée par les 2 premières composantes : {explained_variance_ratio[:2].sum():.2f}")
print(f"Variance cumulée expliquée par les 3 premières composantes : {explained_variance_ratio[:3].sum():.2f}")
```

### Visualisation de la variance expliquée (Scree Plot)

Le Scree Plot nous aide à déterminer le nombre optimal de composantes principales à retenir. Nous cherchons un "coude" où la variance expliquée commence à diminuer de manière moins abrupte.

```python
# Créer un Scree Plot
fig = plt.figure(figsize=(10, 6))
plt.plot(range(1, len(explained_variance_ratio) + 1), explained_variance_ratio, marker='o', linestyle='--')
plt.title('Scree Plot - Variance expliquée par Composante Principale')
plt.xlabel('Numéro de Composante Principale')
plt.ylabel('Proportion de Variance Expliquée')
plt.grid(True)
plt.show()

# Afficher la variance cumulée
cumulative_variance = np.cumsum(explained_variance_ratio)
fig = plt.figure(figsize=(10, 6))
plt.plot(range(1, len(cumulative_variance) + 1), cumulative_variance, marker='o', linestyle='-')
plt.title('Variance Cumulée Expliquée')
plt.xlabel('Nombre de Composantes Principales')
plt.ylabel('Variance Cumulée Expliquée')
plt.grid(True)
plt.show()
```

## Analyse Multivariée : PERMANOVA (Permutational Multivariate Analysis of Variance)

Le PERMANOVA (Analyse de Variance Multivariée par Permutation) est un test non paramétrique utilisé pour détecter les différences significatives entre les centroïdes de groupes dans un espace multivarié. Il est particulièrement utile lorsque les hypothèses des tests paramétriques (comme la MANOVA) ne sont pas satisfaites, ou lorsque l'on travaille avec des matrices de distances. Nous allons l'appliquer aux données numériques standardisées, en utilisant les groupes 'site_variete'.

```python
# Installer la bibliothèque vegpy pour PERMANOVA si ce n'est pas déjà fait
# !pip install vegpy
# Note: la bibliothèque 'vegpy' est ancienne et peut avoir des dépendances conflictuelles ou ne pas être directement installable sur Colab.
# Une alternative plus moderne et largement utilisée est 'skbio' ou 'vegan' (via rpy2).
# Pour simplifier, nous allons utiliser 'skbio' qui offre une fonction PERMANOVA.
!pip install scikit-bio
```

```python
from skbio.stats.distance import permanova, DistanceMatrix
from scipy.spatial.distance import pdist

# S'assurer que df_pca est bien défini à partir de la PCA précédente
# df_pca contient les lignes avec des données complètes après gestion des NaN dans PCA

# La variable X_scaled contient les données numériques standardisées et sans NaN utilisées pour la PCA.
# Calculer la matrice de distance euclidienne à partir des données standardisées.
# PERMANOVA fonctionne sur une matrice de distance.
condensed_distance_matrix = pdist(X_scaled, metric='euclidean')
distance_matrix_skbio = DistanceMatrix(condensed_distance_matrix, ids=df_pca.index)

# Obtenir la variable de groupement 'site_variete' pour les lignes utilisées dans X_scaled
grouping = df_pca['site_variete']

# Exécuter le test PERMANOVA
# Le paramètre 'permutations' peut être augmenté pour plus de précision (par exemple, 9999)
permanova_results = permanova(distance_matrix_skbio, grouping, permutations=999)

print("Résultats du test PERMANOVA :")
display(permanova_results)
```

*   **p-value :** La p-value associée au test. Une p-value faible (typiquement < 0.05) indique une différence significative entre les centroïdes des groupes 'site_variete' dans l'espace multivarié. Cela signifie que la combinaison spécifique du site et de la variété a un effet significatif sur l'ensemble des variables numériques.
*   **Number of permutations :** Le nombre de permutations utilisées pour calculer la p-value empirique. Un nombre plus élevé (par exemple, 9999) rend la p-value plus fiable.
